```python
#Importando as bibliotecas
import pandas as pd
from sqlalchemy import create_engine
from getpass import getpass
from googledrivedownloader import download_file_from_google_drive as gdd
```

```python
# Baixar o arquivo do google drive:
data_google_id = "1NFWiPfKlNt7nMSXPml5Ef7SlP0YkEjL6"
gdd(file_id=data_google_id, dest_path="./dados.csv", showsize=True, overwrite=True)
df = pd.read_csv("./dados.csv")
```

```python
usuario = getpass("Usuário: ")
senha = getpass("Senha: ")
endpoint = getpass("EndPoint: ")
```

```python
url = f"mysql+pymysql://{usuario}:{senha}@{endpoint}:3306/mysql"
print(url)
```

```python
engine = create_engine(url)
with engine.begin() as conexao:
  print("Conexão realizada com sucesso!")
```

```python
with engine.begin() as conexao:
  conexao.exec_driver_sql(
      "CREATE DATABASE api_pi"
      )
  print("Banco de dados criado com sucesso!")
```

```python
# Criando uma tabela manualmente
dados = pd.DataFrame(df)
dados.head()
```

```python
url_database_api_pi = f"mysql+pymysql://{usuario}:{senha}@{endpoint}:3306/api_pi"
engine = create_engine(url_database_api_pi)
dados.to_sql("sensores", engine, if_exists="replace", index=False)
```
